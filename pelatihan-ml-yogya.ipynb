```python
import numpy as np
```

```python
arr_1d = np.array([1,2,3,4,5])
print("Array 1D:",arr_1d,arr_1d.dtype)
```

```python
zeros = np.zeros([1,2,3])
ones = np.ones([3,3])
print("\nArray Zeros:\n", zeros)
print("\nArray Ones:\n",ones)
```

```python
arr_range = np.arange(0, 10, 1.5)
print("\nArray Arange: ", arr_range)
```

```python
arr_lin = np.linspace(0,1,5)
print("\nArray Linspace:",arr_lin)
```

```python
sample_arr = np.array([[1.5,2.3, 4.0],[5.1,6.2,7.3]])
print("Array:")
print(sample_arr)
```

```python
print("Tipe Data (dtype):", sample_arr.dtype)
print("Dimensi (ndin):", sample_arr.ndim)
print("Ukuran / shape:", sample_arr.shape)
print("Total Elemen (size):", sample_arr.size)
```
