```python
nilai = np.array([
    [85,90,88],
    [70,80,75],
    [95,95,100],
    [60,65,70],
    [80,85,80]
])

rata_mapel = np.mean(nilai, axis=0)
print("Rata-rata Nilai (Matematika, Fisika, Kimia):")
print(rata_mapel)
```

```python
rata_siswa = np.mean(nilai,axis=1)
print("Rata-rata nilai per Siswa (A,B,C,D,E):")
print(rata_siswa)
```

```python
nilai_tertinggi = np.max(nilai, axis=0)
print("\nNilai tertinggi tiap Mapel:", nilai_tertinggi)
```

```python
#data sampel : Dadtaset tinggi badan (cm) dan berat badan (kg) dari 5 orang
data = np.array([
    [175, 70],
    [160, 55],
    [180, 85],
    [165, 60],
    [170, 68]
])
```

```python
#Cari nilai minimum dan maksimum untuk setiap kolom
data_min = np.min(data, axis=0)
data_max = np.max(data, axis=0)

#melakukan Min-Max scalling (Broadcasting secara otomatis bekerja di sini)
data_scaled = (data-data_min) / (data_max-data_min)
```

```python
print("Data Asli (Tinggi , Berat):\n", data)
print("\nNilai Min tiap kolom: ", data_min)
print("Nilai Max tiap kolom:", data_max)
print("\nData setelah dinormalisasi (Skala 0-1):\n", np.round(data_scaled,4))
```

# misalnya kita memiliki data volume penjualan harian selama 10 hari. Kita ingin menyaring hari-hari diatas rata dan menghitung total penjualan pada hari-hari tersebut

```python
#volumen penjualan selama 30 hari
penjualan = np.array([100, 85, 150, 200, 95, 100, 75, 210, 130])
hari = np.arange(1,10)
```

```python
#1. Hitung rata-rata penjualan
rata_penjualan = np.mean(penjualan)
print(f"rata-rata penjualan harian: {rata_penjualan} data")
```

```python
#2. Buat boolean mask untuk mendeteksi penjualan diatas rata-rata
mask_di_atas_rata = penjualan > rata_penjualan
print("\nBoolean Mask (Penjualan > Rata-rata):\n", mask_di_atas_rata)
```

```python
#3 Filter data menggunakan mask tersebut
hari_performa_tinggi = hari[mask_di_atas_rata]
penjualan_performa_tinggi = penjualan[mask_di_atas_rata]

print(f"\nHari-hari dengan penjualan diatas rata-rata: hari ke-{hari_performa_tinggi}")
```

```python
print(f"\nNilai penjualan pada hari tersebut: {penjualan_performa_tinggi}")
print(f"\nTotal penjualan yang didapat dari hari-hari performa tinggi tersebut:{np.sum(penjualan_performa_tinggi)}unit")
```
